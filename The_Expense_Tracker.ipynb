```python
# Expense Tracker.py

# Goal: Manage a collection of expense data and generate reports.

expenses = []

def main():
      while True:
        print("\n--- Expense Tracker ---")
        print("1. Add New Expense")
        print("2. Get Totals")
        print("3. Exit")

        choice = input("Selection: ")

        if choice == '1':
            category = input("Enter Category: ")
            vendor = input("Enter Vendor: ")

            # Error handling for non-numeric amounts
            while True:
                try:
                    amount = float(input("Enter Amount: "))
                    break
                except ValueError:
                    print("Invalid input. Please enter a number for the amount.")

        date = input("Enter Date: ")

        expenses.append({
            "category": category.lower(),
            "vendor": vendor.lower(),
            "amount": amount,
            "date":  date
         })

      if choice == '2':
         print("\n--- Totaling Options ---")
         print("A: All Expenses")
         print("C: Filter by Category")
         print("V: Filter by Vendor")
         sub_choice = input("Choice: ").upper()

         total = 0.0

         if sub_choice == 'A':
                total = sum(item['amount'] for item in expenses)
                print(f"The total for all expenses is: ${total:.2f}")

         elif sub_choice == 'C':
                target = input("Enter Category to sum: ").lower()
                total = sum(item['amount'] for item in expenses if item['category'] == target)
                print(f"The total for {target.capitalize()} is: ${total:.2f}")

         elif sub_choice == 'V':
                target = input("Enter Vendor to sum: ").lower()
                total = sum(item['amount'] for item in expenses if item['vendor'] == target)
                print(f"The total for {target.capitalize()} is: ${total:.2f}")

      if choice == '3':
        print("Exiting...")
        'break'

if __name__ == "__main__":
    main()
```
