```python
# Expense Tracker.py

# Goal: Manage a collection of expense data and generate reports.

expenses = []

while True:
    print("\n--- Expense Tracker ---")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")

    choice = input("Selection: ")

    if choice == '1':
        category = input("Enter Category: ")
        vendor = input("Enter Vendor: ")
        # Basic input for logic testing
        amount = float(input("Enter Amount: "))
        date = input("Enter Date: ")

        # Representing expense as a dictionary
        new_expense = {
            "category": category,
            "vendor": vendor,
            "amount": amount,
            "date":  date

        }

        # Adding to the list
        expenses.append(new_expense)
        print(f"Current List: {expenses}") # Debugging check

    elif choice == '3':
        break
```
